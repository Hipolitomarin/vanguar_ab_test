```python
import pandas as pd

# Define the correct process order
CORRECT_ORDER = ['start', 'step_1', 'step_2', 'step_3', 'confirm']
STEP_RANKS = {step: idx for idx, step in enumerate(CORRECT_ORDER)}

def identify_process_errors(df):
    """
    Identifies errors in the process flow for each visit_id:
    1. Consecutive duplicate steps
    2. Out-of-sequence steps
    3. Invalid steps
    4. Premature reset to 'start' without 'confirm'
    5. Proper sequence validation
    """
    df = df.copy()
    df['date_time'] = pd.to_datetime(df['date_time'])
    df_sorted = df.sort_values(['visit_id', 'date_time'])
    df_sorted['error'] = False

    for (visit_id,), group in df_sorted.groupby(['visit_id']):
        previous_step = None
        current_rank = -1
        reached_confirm = False

        for idx, row in group.iterrows():
            step = row['process_step']

            # Case 1: Consecutive duplicate step
            if step == previous_step:
                df_sorted.at[idx, 'error'] = True
                continue

            # Case 2: Premature reset to start
            if step == 'start' and previous_step not in [None, 'confirm']:
                df_sorted.at[idx, 'error'] = True
                current_rank = 0
                previous_step = 'start'
                continue

            # Case 3: Valid start after confirm/None
            if step == 'start':
                current_rank = 0
                reached_confirm = False
                previous_step = 'start'
                continue

            # Case 4: Invalid step
            if step not in STEP_RANKS:
                df_sorted.at[idx, 'error'] = True
                previous_step = step
                continue

            step_rank = STEP_RANKS[step]

            # Case 5: Out of sequence
            if step_rank < current_rank:
                df_sorted.at[idx, 'error'] = True
            else:
                current_rank = step_rank
                if step == 'confirm':
                    reached_confirm = True

            previous_step = step

    return df_sorted.sort_index()
# Process the data
df_validated = identify_process_errors(df_web_experiment_combined)
```

```python
# Error Rates: If there's a step where users go back to a previous step, it may indicate confusion or an error.

error_rates = df_ab.groupby("Variation")["error"].mean()
print("Error rate by group:\n", error_rates)
# The test group showed a higher error rate
```

```python
# Hypothesis for Two-Proportion Z-Test
# p1 = error rate in Test group

# p2 = error rate in Control group

# Null Hypothesis (H₀):
# p1 = p2  — The error rate is the same in both groups.
# Alternative Hypothesis (H₁) (two-sided):
# p1 ≠ p2 
# Alph = 0.05
# Since we are comparing You're comparing proportions (i.e., error rate = mean of error column in group) and our variable is boolean so we will use the Z-test 
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Step 1: Count errors and totals for each group
error_counts = df_ab.groupby("Variation")["error"].sum()
total_counts = df_ab.groupby("Variation")["error"].count()

# Step 2: Run the two-proportion z-test
count = error_counts.values      # number of errors
nobs = total_counts.values       # number of observations
stat, pval = proportions_ztest(count, nobs)

# Step 3: Print results
print(f"Z-statistic: {stat:.3f}")
print(f"P-value: {pval:.3f}")

# Step 4: Interpret
if pval < 0.05:
    print("❗️Statistically significant difference in error rates between Test and Control.")
else:
    print("✅ No statistically significant difference in error rates.")
```

```python
# Ensure 'date_time' column is datetime type
df_sorted = df_validated.copy()
df_sorted['date_time'] = pd.to_datetime(df_sorted['date_time'])

# Sort by visit_id and date_time
df_sorted = df_sorted.sort_values(['visit_id', 'date_time'])

# Add step flags
df_sorted['is_start'] = df_sorted['process_step'] == 'start'
df_sorted['is_confirm'] = df_sorted['process_step'] == 'confirm'

# Prepare list to collect full valid groups
valid_groups = []

# Group by visit_id
for visit_id, group in df_sorted.groupby('visit_id'):
    group = group.reset_index(drop=True)

    # Skip visits with any error
    if group['error'].any():
        continue

    steps = group['process_step'].tolist()
    
    # Only proceed if both 'start' and 'confirm' exist
    if 'start' in steps and 'confirm' in steps:
        start_time = group[group['is_start']].iloc[0]['date_time']
        confirm_time = group[group['is_confirm']].iloc[-1]['date_time']

        # Swap if times are in the wrong order
        if confirm_time < start_time:
            start_time, confirm_time = confirm_time, start_time

        # Add timing info to entire group
        group['start_time'] = start_time
        group['end_time'] = confirm_time
        group['duration'] = confirm_time - start_time

        valid_groups.append(group)

# Combine all valid groups into one DataFrame
df_durations = pd.concat(valid_groups, ignore_index=True)
```

```python
# Filter rows where error is False
filtered_df = df_durations[df_durations['error'] == False]

# Group by Variation and calculate mean duration
average_durations = filtered_df.groupby('Variation')['duration'].mean()

print(average_durations)
```

```python
# Null Hypothesis (H₀):
# There is no significant difference in the average duration for the successful process between the Test and Control groups.
# H0:μTest=μControl
# Alternative Hypothesis (H₁):
# There is a significant difference in the average duration successful process between the Test and Control groups.
# H1:μTest≠μControl
from scipy.stats import ttest_ind

# Filter data (error=False)
test_durations = df_durations[(df_durations['Variation'] == 'Test') & (df_durations['error'] == False)]['duration'].dt.total_seconds()
control_durations = df_durations[(df_durations['Variation'] == 'Control') & (df_durations['error'] == False)]['duration'].dt.total_seconds()

# Perform t-test
t_stat, p_value = ttest_ind(test_durations, control_durations, equal_var=True)

print(f"T-statistic: {t_stat}, P-value: {p_value}")
if p_value < 0.05:
    print("Reject H₀: Significant difference exists.")
else:
    print("Fail to reject H₀: No significant difference.")
```
