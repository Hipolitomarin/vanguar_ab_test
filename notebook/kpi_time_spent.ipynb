```python
import pandas as pd
df = pd.read_csv('../Data/clean/error_rate.csv')
```

```python
# Convert the 'date_time' column to datetime 
df['date_time'] = pd.to_datetime(df['date_time'], format='%Y-%m-%d %H:%M:%S')

# Sort the DataFrame by time and visit id to ensure chronological order
df = df.sort_values(by=['visit_id', 'date_time'])

# Calculate the time differences between consecutive steps within each 'visit_id'
df['duration_between_steps'] = df.groupby('visit_id')['date_time'].diff().shift(-1)

df.head(20)
```

```python
df.info()
```

```python
df = df.dropna(subset=['duration_between_steps'])
```

```python
df.info()
```

```python
df["process_step"].value_counts()
```

```python
df['duration_between_steps'] = pd.to_timedelta(df['duration_between_steps'])
```

```python
df['minutes_decimal'] = df['duration_between_steps'].dt.total_seconds() / 60
df.head(10)
```

```python
average_duration = df.groupby('process_step')['minutes_decimal'].mean()
average_duration
```

```python
average_duration = df.groupby('process_step')['duration_between_steps'].mean()
average_duration
```

```python
test_group = df[df['Variation'] == 'Test']
average_duration_test = test_group.groupby('process_step')['minutes_decimal'].mean()
average_duration_test
```

```python
control_group = df[df['Variation'] == 'Control']
average_duration_control = control_group.groupby('process_step')['minutes_decimal'].mean()
average_duration_control
```

```python
df.to_csv("../Data/clean/kpi_time_spent.csv", index=False, encoding="utf-8", sep=",")
```
